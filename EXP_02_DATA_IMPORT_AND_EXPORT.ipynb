```python
conn = sqlite3.connect('my_database.db')

# Initialize sample employees table if it doesn't exist
conn.execute("""
CREATE TABLE IF NOT EXISTS employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT
)
""")
if pd.read_sql_query("SELECT COUNT(*) as count FROM employees", conn).iloc[0]['count'] == 0:
    conn.execute("INSERT INTO employees (name, department) VALUES ('Alice', 'HR'), ('Bob', 'Engineering')")
    conn.commit()

sql_df = pd.read_sql_query(
    "SELECT * FROM employees",
    conn
)

print("\nSQL Data:")
print(sql_df)
```

```python
import io

url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"}

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.content, "html.parser")

table = soup.find("table", {"class": "wikitable"})
# Wrap the HTML string inside StringIO to support pandas 2.x+
web_df = pd.read_html(io.StringIO(str(table)))[0]

print("\nWeb Data:")
print(web_df.head())
```
