```python
name = "Dharani"
marks = [85, 90, 78, 92, 88]

total = sum(marks)
average = total / len(marks)

print("Student Name:", name)
print("Marks:", marks)
print("Total Marks:", total)
print("Average Marks:", average)
```

```python
name_box = widgets.Text(
      description="Name:",
          placeholder="Enter your name"
          )
display(name_box)


```

```python
button = widgets.Button(
      description="Submit",
          button_style="success"
          )

output = widgets.Output()

def button_clicked(b):
    with output:
        clear_output()
        print("Hello,", name_box.value)
        print("Welcome to Jupyter Widgets!")

button.on_click(button_clicked)

display(button)
display(output)
```

```python
slider = widgets.IntSlider(
      value=50,
          min=0,
              max=100,
                  step=1,
                      description="Marks:"
                      )

display(slider)

```

```python
slider_output = widgets.Output()

def show_marks(change):
    with slider_output:
            clear_output()
            print("Selected Marks:", change["new"])

slider.observe(show_marks, names="value")

display(slider_output)
```

```python
numbers = [10, 20, 30, 40, 50]

mean = sum(numbers) / len(numbers)

print("\nAI-Assisted Python Example")
print("Numbers:", numbers)
print("Mean:", mean)
```
