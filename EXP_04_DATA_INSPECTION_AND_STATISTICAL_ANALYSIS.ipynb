```python
import pandas as pd

# Creating a sample 'expt.csv' file so the code can execute without errors.
# If you have your own dataset, upload it to Colab and replace this block.
with open('expt.csv', 'w') as f:
    f.write('id,value\n1,10\n2,20\n3,30\n')

df = pd.read_csv('expt.csv')
df.head()
```

```python
print(df.head())
print(df.tail())
print(df.info())
print(df.describe())
print(df.columns)
```

```python
categorical_cols = ['id', 'value']

for col in categorical_cols:
    print(f"\nUnique values in '{col}':")
    print(df[col].unique())
```

```python
if 'Dependents' in df.columns:
    df['Dependents'] = df['Dependents'].replace('3+', 3)
    df['Dependents'] = pd.to_numeric(df['Dependents'], errors='coerce').astype('Int64')
    print(df['Dependents'].value_counts(dropna=False))
else:
    print("Warning: 'Dependents' column is not present in the current DataFrame.")
```

```python
cols = ['ApplicantIncome', 'CoapplicantIncome', 'LoanAmount']

# Ensure we only calculate stats for columns that actually exist in the DataFrame
available_cols = [col for col in cols if col in df.columns]
if not available_cols:
    print("Warning: None of the target columns exist in the DataFrame. Using 'value' instead.")
    available_cols = ['value']

for col in available_cols:
    print(f"\nStatistics for '{col}':")
    print(f"Mean: {df[col].mean():.2f}")
    print(f"Median: {df[col].median():.2f}")
    print(f"Mode: {df[col].mode().values}")
    print(f"Range: {df[col].max() - df[col].min():.2f}")
    print(f"Variance: {df[col].var():.2f}")
    print(f"Standard Deviation: {df[col].std():.2f}")
```
